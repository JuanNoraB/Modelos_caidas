```python
from ultralytics import YOLO
from huggingface_hub import hf_hub_download
import os

model_path = hf_hub_download(repo_id="melihuzunoglu/human-fall-detection", filename="best.pt")

model = YOLO(model_path)
 
```

```python
# test input video
import cv2
import matplotlib.pyplot as plt

for i, r in enumerate(model.predict("fall-01-cam1.mp4", conf=0.25, stream=True, verbose=False)):
    if len(r.boxes):
        print(f"frame {i}:", [(model.names[int(b.cls)], round(float(b.conf), 2)) for b in r.boxes])
        plt.imshow(cv2.cvtColor(r.plot(), cv2.COLOR_BGR2RGB))
        plt.axis("off")
        plt.show()
```
